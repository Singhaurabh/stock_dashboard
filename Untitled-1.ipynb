```python
import pandas as pd
import numpy as np
```

```python
#loading the data into the json format

import json

with open("{pair_lower}-updates-{today2}.json",'r') as f:
    data = [json.loads(line)for line in f]
```

```python
# extract the important features from the file asks and bids which we further used in our analysis
snap = data[0]

asks = snap['asks']
bids = snap['bids']

#converting the important columns like aska and bids that I required for the analysis and the convert into the dataframe 
asks_df = pd.DataFrame(asks, columns=["price", "quantity"]).astype(float)
bids_df = pd.DataFrame(bids,columns=['price','quantity']).astype(float)
```

```python
bids_df
```

```python
asks_df
```

```python
asks_df.sort_values(by='price',inplace=True)
bids_df.sort_values(by='price',ascending=False,inplace=True)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,10))
plt.plot(asks_df['price'],asks_df['quantity'],label= 'Asks',color='Green')
plt.plot(bids_df['price'],bids_df['quantity'],label = "Bids",color = "Red")
plt.xlabel("Price")
plt.ylabel("Quantity")
plt.title("Order book depth analysis")
plt.legend()
plt.grid(True)
plt.show()
```
